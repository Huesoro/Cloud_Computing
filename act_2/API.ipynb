```python
import json
import requests
import pandas as pd
```

```python
with open('service_uri.json', 'r') as f:
    service_info = json.load(f)

scoring_uri = service_info['scoring_uri']
print(scoring_uri)
```

## 2. Preparar los datos de prueba

```python
prueba = pd.read_csv('prueba.csv', index_col=0)

X_prueba = prueba.drop(columns=['Bankrupt?'])
y_real = prueba['Bankrupt?']

payload = json.dumps({"data": [X_prueba.to_dict(orient='list')]})
print(f'{X_prueba.shape[0]} registros, {X_prueba.shape[1]} features')
```

## 3. Llamar al servicio

```python
headers = {"Content-Type": "application/json"}

response = requests.post(scoring_uri, data=payload, headers=headers)

if response.status_code == 200:
    result = json.loads(response.json())
    print(result)
else:
    print(f'Error {response.status_code}: {response.text}')
```

## 4. Comparar predicción vs valor real

```python
comparacion = pd.DataFrame({
    'Bankrupt? (real)': y_real.values,
    'Bankrupt? (predicción)': result['result']
})
comparacion
```
