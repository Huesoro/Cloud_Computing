Se esperan 4 campos:

CustomerId(int), Titulo(str), Sufijo (str) y persona de ventas (str)

Titulos aceptados = ["Ms.", "Sr.", "Sra."]

Sufijos aceptados = ["IV", "Jr.", "PhD", "Sr."]

Empleados de venta = ["garrett1", "jae0", "jillian0", "jose1", "linda3", "michael9", "pamela0", "shu0"]:

```python
url = "urlentrega/predecir"

customer_id = 30
titulo = "Ms."
sufijo = "IV"
p_ventas = "jillian0"

data = {"CustomerID": customer_id, "Title": titulo, "Suffix": sufijo, "SalesPerson": p_ventas}
# Hacer el pedido
response = requests.post(url, json=data)

print(response.json())
```
